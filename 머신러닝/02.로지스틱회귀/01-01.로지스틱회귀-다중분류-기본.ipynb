```python
import pandas as pd

df = pd.DataFrame({
    'study_time': [1, 2, 3, 4, 5, 6, 7, 8, 9],
    'grade':      [0, 0, 0, 1, 1, 1, 2, 2, 2]
})

df
```

```python
from sklearn.linear_model import LogisticRegression

# model = LogisticRegression(
#     multi_class='multinomial',  # Softmax 사용
#     solver='lbfgs'
# )

model = LogisticRegression()

X = df[['study_time']]
y = df['grade']

model.fit(X, y)

pred = model.predict(X)
print(pred)
```
