
배깅(Bootstrap Aggregating)은

> **데이터를 여러 번 랜덤하게 뽑아서 여러 모델을 만들고 결과를 평균(투표)하는 방법**

대표 모델:
- 랜덤포레스트(Random Forest)
- BaggingClassifier

# 핵심 정리

- 데이터를 여러 번 복원추출(Bootstrap)
- 여러 모델 생성
- 분류 → 투표
- 회귀 → 평균
- 과적합 감소 효과

## 배깅에서 핵심은:

#### 행(Row)을 랜덤 복원추출
#### 여러 모델 병렬 학습
#### 분류 → 투표
#### 회귀 → 평균
#### 과적합 감소

````python
import pandas as pd

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier

# 데이터
df = pd.DataFrame({
    "study":[1,2,3,4,5,6,7,8],
    "pass":[0,0,0,0,1,1,1,1]
})

X = df[["study"]]
y = df["pass"]

# 기본 모델
tree = DecisionTreeClassifier()

# 배깅 적용
model = BaggingClassifier(
    estimator=tree,
    n_estimators=5,
    random_state=42
)

# 학습
model.fit(X,y)

# 예측
pred = model.predict([[4.5]])

print(pred)
````
