불합격      불합격      합격      합격
   ●-----------●----★----●-----------●
   1           2   2.5   3           4

★ = 예측할 학생

```python
from sklearn.neighbors import KNeighborsClassifier

X = [[1], [2], [3], [4]]
y = [0, 0, 1, 1]  # 0=불합격, 1=합격

model = KNeighborsClassifier(n_neighbors=3)

# 근데 KNN은 다른 모델처럼 수식을 만드는 게 아니라
# 그냥 데이터를 저장해둠.
model.fit(X, y)


result = model.predict([[2.5]])

print(result)
```
