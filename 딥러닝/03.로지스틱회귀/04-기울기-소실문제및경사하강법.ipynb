```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(-10,10,0.1)

y=1/(1+np.exp(-x))

plt.plot(x,y)
plt.grid()
plt.show()
```

```python
# 깊은 신경망에서 기울기 소실 확인
from keras.models import Sequential
from keras.layers import Dense

model = Sequential()

model.add(Dense(100,activation='sigmoid',input_dim=10))

for i in range(10):
    model.add(
        Dense(
            100,
            activation='sigmoid'
        )
    )

model.add(Dense(1))

model.summary()
```
