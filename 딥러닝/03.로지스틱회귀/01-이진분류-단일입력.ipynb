```python
#텐서플로의 케라스 API에서 필요한 함수들을 불러 옵니다.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense , Input

import numpy as np
import matplotlib.pyplot as plt

# 광고비 + 영원사원수 
X = np.array([
    [10,2],
    [20,3],
    [30,4],
    [40,5],
    [50,6]
])

y = np.array([
    100,
    150,
    200,
    250,
    300
])

model = Sequential()


model = Sequential([
    Input(shape=(2,)), # feature 수 => 컬럼수 
    Dense(1,activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


model.fit(
    X,
    y,
    epochs=200,
)

pred = model.predict(np.array([[60,7]]))

print(pred)
```
