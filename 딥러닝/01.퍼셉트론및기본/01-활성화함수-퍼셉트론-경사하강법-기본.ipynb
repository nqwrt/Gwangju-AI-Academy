```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 데이터
X=np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])

y=np.array([
    50,
    60,
    70,
    85,
    95
])

# 학습
model=LinearRegression()

model.fit(
    X,
    y
)

# 예측
pred = model.predict([[6]])

print("6시간 공부 예상 점수")
print(pred)

# 시각화
plt.scatter(X,y)

plt.plot(
    X,
    model.predict(X)
)

plt.show()
```

```python
# y = x₁w₁ + x₂w₂ + x₃w₃ + b

# | 요소     | 의미          |
# | ------ | ----------- |
# | input  | 들어오는 정보     |
# | weight | 얼마나 중요하게 볼지 |
# | bias   | 기준값 조정      |
# | output | 최종 판단       |


# 확장
inputs = [1,2,3]
weights = [0.3, 0.5, 0.2]
bias = 1
output = 0

for i in range(len(inputs)):
    output += inputs[i] * weights[i]    

output += bias
print(output)

```

```python
# 넘파이 버전

import numpy as np

inputs = np.array([1,2,3])
weights = np.array([0.3,0.5,0.2])
bias = 1

# 내적(dot product)
output = np.dot(
    inputs,
    weights
) + bias

print(output)
```

```python
# 입력:
# 입력:
# 평수 = 30 ,20 ,50
# 역거리 = 10 , 5 , 10
# 연식 = 5 , 4, 3

# # 가중치
# 평수 → 0.8
# 역거리 → -0.3
# 연식 → -0.1

# 각가의 집값을 계산하시오 => 넘파이를 활용하여 하시오

import numpy as np

X = np.array([
[30,10,5],
[20,5,4],
[50,10,3]
])

W = np.array([
0.8,
-0.3,
-0.1
])

output = X @ W

print(output)
```

```python
# 활성화 함수의 이해

# Step
# =
# 꺼짐 / 켜짐

# Sigmoid
# =
# 0~1 사이 확률처럼 출력

# 활성화 함수
# =
# 뉴런에 비선형성을 추가해서
# 복잡한 문제를 풀게 만드는 장치

def step(x):
    if x>=0:
        return 1
    return 0


for i in [-5,-1,0,1,5]:
    print(i,step(i))

def sigmoid(x):

    return (
        1/
        (
            1+
            np.exp(-x)
        )
    )

# 0 ─────┐
#        │
# 1      └──────

x = np.arange(-10,10,0.1)
y = sigmoid(x)

plt.plot(x,y)
plt.show()
```

```python
def perceptron(x1,x2):
    
    w1=1
    w2=1
    b=-1.5
    
    y = x1*w1 + x2*w2 + b

    if y > 0 :
        return 1

    return 0


data=[
(0,0),
(0,1),
(1,0),
(1,1)
]

for x1,x2 in data:

    print(
        x1,
        x2,
        perceptron(
            x1,
            x2
        )
    )
```

```python
from sklearn.linear_model import Perceptron

X=[
    [0,0],
    [0,1],
    [1,0],
    [1,1]
]

y = [0,1,1,0]

model = Perceptron()

# 내부동작

# 초기 상태

# w=랜덤
# b=랜덤
# w=[0.1 0.2]
# b=0

# 학습

# 입력
# ↓
# 예측
# ↓
# 정답 비교
# ↓
# 오차 계산
# ↓
# 가중치 수정
# 계속 반복

# 결국 이런 식의 뉴런을 찾음
# y = x₁w₁+x₂w₂+b
# 예)
# w=[1,1]
# b=-1.5

model.fit(
    X,
    y
)

print("가중치:")
print(model.coef_)

print()

print("편향:")
print(model.intercept_)

pred = model.predict(X)

print(pred)

# 예측 실패
# ↓

# XOR 해결 불가
# ↓

# 은닉층 필요
```

```python
# 8교시 — 경사하강법 직접 구현

# 오차 계산
# ↓

# 기울기 계산
# ↓

# 가중치 수정
# ↓

# 반복
w = 10
lr = 0.1

for epoch in range(20):

    loss = (w-3)**2    
    grad = 2 * (w-3) # 위의 미분
    
    #새 가중치 = 기존 가중치 - 학습률×기울기
    w = w - lr * grad 

    print(epoch,round(w,3),round(loss,3))
```
