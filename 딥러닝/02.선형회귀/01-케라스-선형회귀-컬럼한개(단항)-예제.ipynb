```python

#텐서플로의 케라스 API에서 필요한 함수들을 불러 옵니다.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([2, 4, 6, 8])
y = np.array([81, 93, 91, 97]) 
```

```python

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense
model = Sequential()

# 출력 값, 입력 변수, 분석 방법에 맞게끔 모델을 설정합니다. 
# model.add(Dense(1, input_dim=1, activation='linear'))
# model.add(Dense(1, input_shape=(1,), activation='linear'))
# model.add(Dense(1, Input(shape=(1,)), activation='linear'))

# x = np.array([
#     [2],
#     [4],
#     [6],
#     [8]
# ])
#(4,1)
# 데이터 4개
# 특성 1개

model = Sequential([
    Input(shape=(1,)), #입력 특성(feature)이 1개
    Dense(1, activation='linear') #활성화 함수 없음 , f(x)=x, 출력층 , 은닉층 자체는 없는 상태 
])

# 오차 수정을 위해 경사 하강법(sgd)을, 오차의 정도를 판단하기 위해 평균 제곱 오차(mse)를 사용합니다. 
model.compile(optimizer='sgd', loss='mse')

# 오차를 최소화하는 과정을 2000번 반복합니다.
model.fit(x, y, epochs=100)
```

```python
plt.scatter(x, y)
plt.plot(x, model.predict(x),'r')    # 예측 결과를 그래프로 나타냅니다.
plt.show()
```

```python
weights,bias = model.layers[0].get_weights()

print(weights)
print(bias)
model.summary()  
```

```python
# 6. 새로운 데이터로 예측하기
# 예: 공부 시간 7시간, 과외 시간 2시간일 때 성적 예측
new_data = np.array([10,11]) 

predictions = model.predict(new_data)
print(predictions)
```

```python
# 문제 3. 자동차 속도와 연비

# 데이터

# x = np.array([
#     40,
#     60,
#     80,
#     100,
#     120
# ])

# y = np.array([
#     25,
#     22,
#     19,
#     16,
#     13
# ])
# 해야 할 일
# 모델 생성
# 학습 수행
# 속도 90km일 때 연비 예측
```

```python
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

# 데이터
x = np.array([
    40,
    60,
    80,
    100,
    120
])

y = np.array([
    25,
    22,
    19,
    16,
    13
])

# 모델 생성
model = Sequential([
    Input(shape=(1,)),
    Dense(1, activation='linear')
])

# 컴파일
model.compile(
    optimizer='adam',
    loss='mse'
)

# 학습
model.fit(
    x,
    y,
    epochs=1000,
    verbose=0
)

# 가중치와 바이어스 확인
weights, bias = model.layers[0].get_weights()

print("가중치:", weights)
print("바이어스:", bias)

# 속도 90km 예측
speed = 90

prediction = model.predict(
    np.array([[speed]]),
    verbose=0
)

print(f"속도 {speed}km/h 예상 연비: {prediction[0][0]:.2f}")
```
