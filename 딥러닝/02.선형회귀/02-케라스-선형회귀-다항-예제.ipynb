```python
import numpy as np
import matplotlib.pyplot as plt

#텐서플로의 케라스 API에서 필요한 함수들을 불러 옵니다.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

x = np.array([[2, 0], [4, 4], [6, 2], [8, 3]])
y = np.array([81, 93, 91, 97]) 
```

```python
model = Sequential()

#입력 변수가 2개(학습 시간, 과외 시간)이므로 input_dim에 2를 입력합니다. 
model.add(Dense(1, input_dim=2, activation='linear'))
model.compile(optimizer='sgd' ,loss='mse')

model.fit(x, y, epochs=2000)


# 입력층 + 은닉층 1
# model.add(Dense(2, input_dim=1, activation='linear'))

# # 은닉층 2
# model.add(Dense(2, activation='linear'))
# # 출력층
# model.add(Dense(1, activation='linear'))
# model.compile(optimizer='sgd', loss='mse')

# 입력층(1개)
#       ↓

# 은닉층1 (뉴런 2개)
#    ○     ○

#       ↓

# 은닉층2 (뉴런 2개)
#    ○     ○

#       ↓

# 출력층 (뉴런 1개)
#         ○


# model = Sequential([
#     Input(shape=(2,)),          # 입력층
#     Dense(2, activation='linear'),  # 은닉층 1
#     Dense(1, activation='linear')   # 출력층
# ])

# # model.compile(
# #     optimizer='sgd',
# #     loss='mse' # 폭발 함
# # )
# from tensorflow.keras.optimizers import SGD

# model.compile(
#     optimizer = SGD(
#         learning_rate=0.0001
#     ),
#     loss='mse'
# )
# 파라미터는 Param는 모델이 학습하는 숫자의 개수
# 👉 즉, 학습 중에 계속 바뀌는 값들
# 🧠 왜 중요한가?

# Param이 많다는 건:

# ✔ 모델이 더 복잡하다
# ✔ 학습할 게 많다
# ✔ 표현력이 크다

# 하지만 동시에:

# ❌ 과적합 위험 증가
# ❌ 학습 느려짐
model.summary()  

model.fit(x, y, epochs=100)
```

```python
# 다항 문제
# 문제 3. 광고비 + 영업사원수 → 매출

import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

# ==========================
# 데이터
# 광고비, 영업사원수
# ==========================

x = np.array([
    [10, 2],
    [20, 3],
    [30, 4],
    [40, 5]
])

# 매출
y = np.array([
    100,
    180,
    260,
    340
])

# ==========================
# 모델 생성
# ==========================

model = Sequential([
    Input(shape=(2,)),
    Dense(1, activation='linear')
])

# ==========================
# 컴파일
# ==========================

model.compile(
    optimizer='adam',
    loss='mse'
)

# ==========================
# 학습
# ==========================

model.fit(
    x,
    y,
    epochs=3000,
    verbose=0
)

# ==========================
# 가중치 확인
# ==========================

weights, bias = model.layers[0].get_weights()

print("광고비 가중치 :", weights[0][0])
print("영업사원수 가중치 :", weights[1][0])
print("바이어스 :", bias[0])

# ==========================
# 예측
# 광고비=35
# 영업사원수=4
# ==========================

new_data = np.array([
    [35, 4]
])

prediction = model.predict(
    new_data,
    verbose=0
)

print()
print("예상 매출 :", prediction[0][0])
```
